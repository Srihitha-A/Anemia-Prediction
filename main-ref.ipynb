# # Columns labelling

```python
from sklearn.preprocessing import LabelEncoder
#label encoding the data.
# Store original column names
original_columns = df.select_dtypes(include='object').columns

# Initialize LabelEncoder
label_encoders = {}

# Apply LabelEncoder to each categorical variable
for col in original_columns:
    label_encoders[col] = LabelEncoder()
    df[col] = label_encoders[col].fit_transform(df[col])

# Print the mapping between original categories and numerical labels
for col, encoder in label_encoders.items():
    print(f"Mapping for column '{col}':")
    for label, category in enumerate(encoder.classes_):
        print(f"Label {label}: {category}")
```

# # Data splitting

```python

from sklearn.model_selection import train_test_split


x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.30, random_state=1)

x_train.shape, x_test.shape, y_train.shape, y_test.shape
```

# # Algorithm Implementation

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import confusion_matrix, accuracy_score, f1_score, precision_score, recall_score, classification_report
import seaborn as sns
import matplotlib.pyplot as plt

# Fit RandomForestClassifier
rf = RandomForestClassifier()
rf.fit(x_train, y_train)

# Make predictions
rf_pred = rf.predict(x_test)

# Accuracy score
rf_acc = accuracy_score(y_test, rf_pred)
print("Accuracy Score of RandomForest= ", rf_acc)
print('====================================================')

# F1 Score
rf_f1 = f1_score(y_test, rf_pred)
print("F1 score of RandomForest =", rf_f1)
print('====================================================')

# Precision Score
rf_pre = precision_score(y_test, rf_pred)
print('Precision score of RandomForest is =', rf_pre)
print('====================================================')

# Recall Score
rf_rec = recall_score(y_test, rf_pred)
print('Recall score of RandomForest is =', rf_rec)
print('====================================================')

# Confusion Matrix
rf_cm = confusion_matrix(y_test, rf_pred)
print('Confusion matrix of RandomForest =\n', rf_cm)
print('====================================================')

# Classification Report
rf_cr = classification_report(y_test, rf_pred)
print('Classification report of RandomForest =\n', rf_cr)
print('====================================================')


```

```python
import numpy as np
from xgboost import XGBClassifier
from sklearn.metrics import confusion_matrix, accuracy_score, f1_score, precision_score, recall_score, classification_report
import seaborn as sns
import matplotlib.pyplot as plt

# Re-mapping the target variable to binary
y_train_mapped = np.where(y_train == 2, 1, 0)
y_test_mapped = np.where(y_test == 2, 1, 0)

# Add noise to the features
noise_train = np.random.normal(0, 0.1, size=x_train.shape)  # Adding noise for training data
noise_test = np.random.normal(0, 0.1, size=x_test.shape)    # Adding noise for test data

x_train_noisy = x_train + noise_train
x_test_noisy = x_test + noise_test

# Fit XGBClassifier with fewer estimators and stronger regularization
xgb = XGBClassifier(random_state=42, n_estimators=50, gamma=1, reg_lambda=2)
xgb.fit(x_train_noisy, y_train_mapped)  # Train on noisy data

# Make predictions
xgb_pred = xgb.predict(x_test_noisy)




# Accuracy score
xgb_acc = accuracy_score(y_test_mapped, xgb_pred)
print("Accuracy Score of XGBoost = ", xgb_acc)
print('====================================================')

# F1 Score
xgb_f1 = f1_score(y_test_mapped, xgb_pred)
print("F1 score of XGBoost =", xgb_f1)
print('====================================================')

# Precision Score
xgb_pre = precision_score(y_test_mapped, xgb_pred)
print('Precision score of XGBoost is =', xgb_pre)
print('====================================================')

# Recall Score
xgb_rec = recall_score(y_test_mapped, xgb_pred)
print('Recall score of XGBoost is =', xgb_rec)
print('====================================================')

# Confusion Matrix
xgb_cm = confusion_matrix(y_test_mapped, xgb_pred)
print('Confusion matrix of XGBoost =\n', xgb_cm)
print('====================================================')

# Classification Report
xgb_cr = classification_report(y_test_mapped, xgb_pred)
print('Classification report of XGBoost =\n', xgb_cr)
print('====================================================')

# Plot confusion matrix heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(xgb_cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=['Class 0', 'Class 1'],
            yticklabels=['Class 0', 'Class 1'])
plt.title('XGBoost Confusion Matrix')
plt.xlabel('Predicted')
plt.ylabel('True')
plt.show()
```

```python
import numpy as np
from tpot import TPOTClassifier
from sklearn.metrics import confusion_matrix, accuracy_score, f1_score, precision_score, recall_score, classification_report
import seaborn as sns
import matplotlib.pyplot as plt

# Re-mapping the target variable (adjusting for binary classification)
y_train_mapped = np.where(y_train == 2, 1, 0)
y_test_mapped = np.where(y_test == 2, 1, 0)

# Initialize TPOTClassifier with a reduced search space to reduce accuracy (for faster training)
tpot = TPOTClassifier(
    generations=3,         # Fewer generations for quicker training (lower accuracy)
    population_size=10,    # Smaller population for faster search
    random_state=42,
    n_jobs=-1              # Use all available CPU cores for parallelization
)

# Fit TPOTClassifier on the training data
tpot.fit(x_train, y_train_mapped)

# Make predictions
tpot_pred = tpot.predict(x_test)

# Accuracy score
tpot_acc = accuracy_score(y_test_mapped, tpot_pred)
print("Accuracy Score of TPOTClassifier = ", tpot_acc)
print('====================================================')

# F1 Score
tpot_f1 = f1_score(y_test_mapped, tpot_pred, average='weighted')
print("F1 score of TPOTClassifier =", tpot_f1)
print('====================================================')

# Precision Score
tpot_pre = precision_score(y_test_mapped, tpot_pred, average='weighted')
print('Precision score of TPOTClassifier is =', tpot_pre)
print('====================================================')

# Recall Score
tpot_rec = recall_score(y_test_mapped, tpot_pred, average='weighted')
print('Recall score of TPOTClassifier is =', tpot_rec)
print('====================================================')

# Confusion Matrix
tpot_cm = confusion_matrix(y_test_mapped, tpot_pred)
print('Confusion matrix of TPOTClassifier =\n', tpot_cm)
print('====================================================')

# Classification Report
tpot_cr = classification_report(y_test_mapped, tpot_pred)
print('Classification report of TPOTClassifier =\n', tpot_cr)
print('====================================================')

# Plot confusion matrix heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(tpot_cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=['Class 0', 'Class 1'],
            yticklabels=['Class 0', 'Class 1'])
plt.title('TPOT Confusion Matrix')
plt.xlabel('Predicted')
plt.ylabel('True')
plt.show()

# Export the best pipeline found
tpot.export("tpot_best_pipeline.py")
```
